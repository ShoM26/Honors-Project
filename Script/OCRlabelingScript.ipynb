```python
import os
import cv2
import tkinter as tk
from PIL import Image, ImageTk

# --- Configuration ---
SOURCE_FOLDER = 'Annotated_Better_Again/OCRData/TrOCR'  # Folder containing your PNGs
OUTPUT_FOLDER = 'Annotated_Better_Again/OCRData/TrOCR/OCRlabels/'  # Where the txt files will be saved

class ImageLabeler:
    def __init__(self, root):
        self.root = root
        self.root.title("Quick Labeler")
        
        # Ensure output directory exists
        if not os.path.exists(OUTPUT_FOLDER):
            os.makedirs(OUTPUT_FOLDER)

        # Get list of PNGs that don't have a label yet
        self.all_files = [f for f in os.listdir(SOURCE_FOLDER) if f.lower().endswith('.png')]
        self.files_to_label = [f for f in self.all_files if not self._label_exists(f)]
        self.current_index = 0

        if not self.files_to_label:
            print("No new images to label!")
            self.root.destroy()
            return

        # UI Elements
        self.img_label = tk.Label(root)
        self.img_label.pack(pady=10)

        self.label_var = tk.StringVar()
        self.entry = tk.Entry(root, textvariable=self.label_var, width=50)
        self.entry.pack(pady=10, padx=20)
        self.entry.bind("<Return>", self.save_and_next) # Bind Enter key
        self.entry.focus_set()

        self.status_label = tk.Label(root, text="", fg="gray")
        self.status_label.pack()

        self.load_image()

    def _label_exists(self, filename):
        txt_name = os.path.splitext(filename)[0] + ".txt"
        return os.path.exists(os.path.join(OUTPUT_FOLDER, txt_name))

    def load_image(self):
        if self.current_index < len(self.files_to_label):
            filename = self.files_to_label[self.current_index]
            path = os.path.join(SOURCE_FOLDER, filename)
            
            # Read with OpenCV, convert to RGB for Tkinter
            cv_img = cv2.imread(path)
            cv_img = cv2.cvtColor(cv_img, cv2.COLOR_BGR2RGB)
            
            # Resize image to fit screen if necessary (e.g., max 800px width)
            h, w, _ = cv_img.shape
            if w > 800:
                ratio = 800 / w
                cv_img = cv2.resize(cv_img, (800, int(h * ratio)))

            self.img_tk = ImageTk.PhotoImage(image=Image.fromarray(cv_img))
            self.img_label.config(image=self.img_tk)
            self.status_label.config(text=f"Image {self.current_index + 1}/{len(self.files_to_label)}: {filename}")
            self.label_var.set("") # Clear text box
        else:
            print("Finished labeling all images.")
            self.root.destroy()

    def save_and_next(self, event=None):
        label = self.label_var.get().strip()
        if label:
            filename = self.files_to_label[self.current_index]
            txt_name = os.path.splitext(filename)[0] + ".txt"
            save_path = os.path.join(OUTPUT_FOLDER, txt_name)

            with open(save_path, 'w') as f:
                f.write(label)
            
            print(f"Saved: {txt_name}")
            self.current_index += 1
            self.load_image()
        else:
            print("Please enter a label before hitting Enter.")

if __name__ == "__main__":
    root = tk.Tk()
    app = ImageLabeler(root)
    root.mainloop()
```
