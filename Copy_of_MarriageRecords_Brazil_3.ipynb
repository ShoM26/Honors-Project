```python
# ----------- Extraction of the bounding boxes -----------

# Method to crop image
def cropImage(j, crop_name, img, top_left, bottom_right):
    # Cropping an image
    img_crop = img[top_left[j][1]:bottom_right[j][1], top_left[j][0]:bottom_right[j][0]]

    # Ensure the output folder exists
    if not os.path.exists("/content/croppedImages"):
        os.makedirs("/content/croppedImages")

    # Exception handling (not grabbed bounding boxes)
    error_image = cv2.imread("/content/error_handle_pic.jpg")

    if img_crop.any() == False:
      cv2.imwrite(os.path.join("/content/croppedImages", crop_name), error_image)
    else:
      # Save the cropped image to the specified folder
      cv2.imwrite(os.path.join("/content/croppedImages", crop_name), img_crop)

# Defining bounding box
def makeBox(name, j, color, lines, top_left, bottom_right, img):
  index = int(lines[j].split()[0])
  img_d = cv2.rectangle(img, top_left[j], bottom_right[j], color,2)

# Extracting bounding box
def extractBox(number, name, j, color, lines, top_left, bottom_right, img, crop_name):
  if (name == number): # making boxes for milleottocentosettanta
    makeBox(name,j, color,lines, top_left, bottom_right, img)
    cropImage(j,crop_name, img, top_left, bottom_right)

# Define title of bouding box
def title_extract_to_file(number):
  title_file = [file for file in os.listdir("/content/titles_folder") if file.lower().endswith('.jpg')]
  image_files = sorted(image_files)
```

```python
# ----------- Concatinating extraction boxes -----------

# Uniformly resize the extraction boxes
def hconcat_resize(img_list, interpolation=cv2.INTER_CUBIC):
    # Resize images to have the same height (assumes images have the same width)
    img_list_resized = [cv2.resize(img, (img_list[0].shape[1], img_list[0].shape[0]), interpolation=interpolation)
                        for img in img_list]
    # Horizontally concatenate all the images
    return cv2.hconcat(img_list_resized)


# Horizontally concatenate bounding boxes
def h_concatenate_and_save():
    # Get a list of image files in the input folder
    image_files = [file for file in os.listdir("/content/croppedImages") if file.lower().endswith('.jpg')]
    image_files = sorted(image_files)
    # Read images from the folder
    images = [cv2.imread(os.path.join("/content/croppedImages", img)) for img in image_files]
    # Horizontally concatenate all the images
    img_out_put = hconcat_resize(images)
    # Ensure the output folder exists
    output_folder = "/content/hConcatImages"
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    # Generate output file name based on current timestamp to avoid overrides
    timestamp = int(time.time())
    output_file = f"concatenated_{timestamp}.jpg"
    # Save the concatenated image to the output folder
    cv2.imwrite(os.path.join(output_folder, output_file), img_out_put)

def h_concatenate_and_save_for_titles():
    # Get a list of image files in the input folder
    image_files = [file for file in os.listdir("/content/titles_folder") if file.lower().endswith('.jpg')]
    image_files = sorted(image_files)
    # Read images from the folder
    images = [cv2.imread(os.path.join("/content/titles_folder", img)) for img in image_files]
    # Horizontally concatenate all the images
    img_out_put = hconcat_resize(images)
    # Ensure the output folder exists
    output_folder = "/content/vConcatImages"
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    # Generate output file name based on current timestamp to avoid overrides
    timestamp = int(time.time())
    output_file = f"concatenated_{timestamp}.jpg"
    # Save the concatenated image to the output folder
    cv2.imwrite(os.path.join(output_folder, output_file), img_out_put)


# Resize images vertically for uniform rows
def vconcat_resize(img_list, interpolation = cv2.INTER_CUBIC):
    # take minimum width
    w_min = min(img.shape[1] for img in img_list)
    # resizing images
    im_list_resize = [cv2.resize(img,(w_min, int(img.shape[0] * w_min / img.shape[1])),interpolation = interpolation) for img in img_list]
    # return final image
    return cv2.vconcat(im_list_resize)


# Save vertically concatinated images
def v_concatenate_and_save():
    # Get a list of image files in the input folder
    image_files = [file for file in os.listdir("/content/hConcatImages") if file.lower().endswith('.jpg')]
    image_files = sorted(image_files)
    # Read images from the folder
    images = [cv2.imread(os.path.join("/content/hConcatImages", img)) for img in image_files]
    # Vertically concatenate all the images
    img_out_put = vconcat_resize(images)
    # Ensure the output folder exists
    output_folder = "/content/spreadsheet"
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    # Generate output file name based on current timestamp to avoid overrides
    timestamp = int(time.time())
    output_file = f"spreadsheet_{timestamp}.jpg"
    # Save the concatenated image to the output folder
    cv2.imwrite(os.path.join(output_folder, output_file), img_out_put)


# Save final spreadsheet
def final_v_concatenate_and_save():
    # Get a list of image files in the input folder
    image_files = [file for file in os.listdir("/content/spreadsheet") if file.lower().endswith('.jpg')]
    image_files = sorted(image_files)
    # Read images from the folder
    images = [cv2.imread(os.path.join("/content/spreadsheet", img)) for img in image_files]
    # Vertically concatenate all the images
    img_out_put = vconcat_resize(images)
    # Ensure the output folder exists
    output_folder = "/content/spreadsheet"
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    # Generate output file name based on current timestamp to avoid overrides
    timestamp = int(time.time())
    output_file = f"spreadsheet_{timestamp}.jpg"
    # Save the concatenated image to the output folder
    cv2.imwrite(os.path.join(output_folder, output_file), img_out_put)


# Clear contents of hConcatImages folder
def clear_hconcat_images_folder():
    folder_path = "/content/hConcatImages"
    # Check if the folder exists
    if os.path.exists(folder_path):
        # Remove all files in the folder
        for filename in os.listdir(folder_path):
            file_path = os.path.join(folder_path, filename)
            try:
                if os.path.isfile(file_path) or os.path.islink(file_path):
                    os.unlink(file_path)
                elif os.path.isdir(file_path):
                    shutil.rmtree(file_path)
            except Exception as e:
                print(f"Failed to delete {file_path}. Reason: {e}")
    else:
        print(f"'{folder_path}' folder does not exist.")


# Clear contents of vConcatImages folder
def clear_vconcat_images_folder():
    folder_path = "/content/spreadsheet"
    # Check if the folder exists
    if os.path.exists(folder_path):
        # Remove all files in the folder
        for filename in os.listdir(folder_path):
            file_path = os.path.join(folder_path, filename)
            try:
                if os.path.isfile(file_path) or os.path.islink(file_path):
                    os.unlink(file_path)
                elif os.path.isdir(file_path):
                    shutil.rmtree(file_path)
            except Exception as e:
                print(f"Failed to delete {file_path}. Reason: {e}")
    else:
        print(f"'{folder_path}' folder does not exist.")
```
